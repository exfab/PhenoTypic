# How To: Correct Color Cast Across Plates

Apply `ColorCorrector` with a fitted `ColorCheckerProfile` to standardize
color across plates imaged under different lighting conditions.

```python
import phenotypic as pht
from phenotypic.data import load_yeast_plate_full
from phenotypic.correction import ColorCorrector, ColorCheckerProfile
```

## Load a Full Plate with Color Checker

Color correction requires an image that includes a color checker target.
The full (uncropped) plate images include one.

```python
plate = load_yeast_plate_full()
plate.dash()
```

## Fit a Color Checker Profile

Create a `ColorCheckerProfile` and fit it to the color checker region
in your image. This learns the mapping from observed to reference colors.

```python
profile = ColorCheckerProfile()
profile.fit(plate)
```

## Apply the Correction

```python
corrector = ColorCorrector(profile=profile)
corrected = corrector(plate)
corrected.dash()
```

The corrected image has standardized colors. Apply the same fitted
`profile` to all plates from the same imaging session for consistent
color measurements across your dataset.
